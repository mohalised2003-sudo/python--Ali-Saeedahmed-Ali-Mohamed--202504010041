# Week 3 Tutorial: Student Average Calculator

## 1. Identify the Components

### 1.1 Inputs
* `quiz_1`: Mark for Quiz 1.
* `quiz_2`: Mark for Quiz 2.
* `quiz_3`: Mark for Quiz 3.
* `choice`: User input (`y` or `n`) to decide whether to continue entering marks for another student.

### 1.2 Process
* Calculate the average of the three quiz marks: `average = (quiz_1 + quiz_2 + quiz_3) / 3`.
* Check if `average >= 50` to determine whether the student passes or fails.
* Repeat the process using a `while` loop as long as `choice == "y"`.

### 1.3 Output
* Display the calculated `average` mark.
* Display the result (`Pass` or `Fail`).
* Display `Program Ended` when the loop terminates.

## 2. Design the Algorithm

### 2.1 Pseudocode
```text
START
    SET choice = "y"
    WHILE choice == "y" DO
        INPUT quiz_1
        INPUT quiz_2
        INPUT quiz_3

        SET average = (quiz_1 + quiz_2 + quiz_3) / 3
        OUTPUT average

        IF average >= 50 THEN
            OUTPUT "Pass"
        ELSE
            OUTPUT "Fail"
        ENDIF

        INPUT choice ("Continue? Select Y/N: ")
    ENDWHILE
    OUTPUT "Program Ended"
END
```

### 2.2 Identify the Algorithm Structure
* **2.2.1 Sequence:** Reading the three quiz marks one after another (`quiz_1`, `quiz_2`, `quiz_3`), calculating the `average`, and printing it.
* **2.2.2 Selection:** Using the `if-else` statement (`if average >= 50:`) to choose between printing `Pass` or `Fail`.
* **2.2.3 Iteration:** Using the `while` loop (`while choice == "y":`) to repeat the steps for multiple students.
